## Q1 — Active revenue

**Answer:** **2,340 active subscriptions** and **₹784,260 total monthly recurring revenue** as of 30 June 2024. The definition correctly includes open-ended subscriptions where `end_date IS NULL`. fileciteturn0file0L164-L179


```python
# Q1: Active revenue

query = """
SELECT
    COUNT(*) AS active_subscription_count,
    SUM(monthly_price_inr) AS total_monthly_revenue_inr
FROM subscriptions
WHERE status = 'active'
  AND (end_date IS NULL OR end_date > '2024-06-30');
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q2 — Signup momentum

**Answer:** January = **350**, February = **400**, March = **500**, April = **550**, May = **600**, June = **600**. **May and June are tied for the highest signups (600 each).** fileciteturn0file0L180-L184


```python
# Q2: Signup momentum

query = """
SELECT
    DATE_FORMAT(signup_date, '%Y-%m') AS month,
    COUNT(*) AS signup_count
FROM users
WHERE signup_date >= '2024-01-01'
  AND signup_date < '2024-07-01'
GROUP BY DATE_FORMAT(signup_date, '%Y-%m')
ORDER BY month;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q3 — Device analytics

**Answer:** The code prints the required device-level session count, total watch minutes, average watch minutes, and completion rate. NULL `user_id` sessions are excluded and completion rate is rounded to 2 decimals, exactly as requested. fileciteturn0file0L185-L191


```python
# Q3: Device analytics

query = """
SELECT
    device_type,
    COUNT(*) AS total_sessions,
    SUM(watch_minutes) AS total_watch_minutes,
    ROUND(AVG(watch_minutes), 2) AS avg_watch_minutes,
    ROUND(
        100.0 * SUM(CASE WHEN completed = 1 THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS completion_rate_pct
FROM watch_sessions
WHERE user_id IS NOT NULL
GROUP BY device_type
ORDER BY device_type;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q4 — Rating distribution

**Answer:** 1★ = **234 (4.68%)**, 2★ = **352 (7.04%)**, 3★ = **847 (16.94%)**, 4★ = **1,781 (35.62%)**, 5★ = **1,786 (35.72%)**. **4★ or 5★ ratings = 71.34%.** fileciteturn0file0L192-L196


```python
# Q4: Rating distribution

query = """
SELECT
    stars,
    COUNT(*) AS rating_count,
    ROUND(
        100.0 * COUNT(*) / (SELECT COUNT(*) FROM ratings),
        2
    ) AS rating_percentage
FROM ratings
GROUP BY stars
ORDER BY stars;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

```python
# Q4: Overall percentage of ratings that are 4 or 5 stars.

four_five_query = """
SELECT ROUND(
    100.0 * SUM(CASE WHEN stars IN (4, 5) THEN 1 ELSE 0 END) / COUNT(*),
    2
) AS four_or_five_star_percentage
FROM ratings;
"""

print(pd.read_sql(four_five_query, engine).to_string(index=False))
```

## Q5 — Originals vs acquired

**Answer:** BingePlay Originals average **7.92 IMDb**, while acquired content averages **6.63 IMDb**. Originals perform better by **1.29 IMDb points**. fileciteturn0file0L197-L206


```python
# Q5: Originals vs acquired

query = """
SELECT
    CASE
        WHEN is_original = 1 THEN 'BingePlay Original'
        ELSE 'Acquired Content'
    END AS content_group,
    COUNT(*) AS number_of_shows,
    ROUND(AVG(imdb_rating), 2) AS avg_imdb_rating,
    ROUND(AVG(release_year), 2) AS avg_release_year
FROM shows
GROUP BY is_original
ORDER BY is_original;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q6 — Binge day detection

**Answer:** **414 binge days** occurred in Q2 2024. **U02956** had the most with **8 binge days**. The grouping uses `(user_id, show_id, session_date)`, matching the definition. fileciteturn0file0L210-L221


```python
# Q6: Binge day detection

query = """
WITH binge_days AS (
    SELECT
        user_id,
        show_id,
        session_date,
        COUNT(*) AS session_count
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date BETWEEN '2024-04-01' AND '2024-06-30'
    GROUP BY user_id, show_id, session_date
    HAVING COUNT(*) >= 5
),
user_binge_counts AS (
    SELECT user_id, COUNT(*) AS binge_days
    FROM binge_days
    GROUP BY user_id
)
SELECT
    (SELECT COUNT(*) FROM binge_days) AS total_binge_days,
    user_id AS top_binge_user,
    binge_days AS top_user_binge_days
FROM user_binge_counts
ORDER BY binge_days DESC, user_id
LIMIT 1;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q7 — Q1 signups who never watched

**Answer:** **1,250 Q1 signups** and **226 Q1 users who never watched**. `NOT EXISTS` is used intentionally to avoid the `NOT IN` + `NULL` trap described in the brief. fileciteturn0file0L222-L235


```python
# Q7: Q1 signups who never watched

query = """
SELECT
    COUNT(*) AS never_watched_q1_users
FROM users u
WHERE u.signup_date BETWEEN '2024-01-01' AND '2024-03-31'
  AND NOT EXISTS (
      SELECT 1
      FROM watch_sessions ws
      WHERE ws.user_id = u.user_id
  );
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

```python
# Q7: Total number of Q1 signups.

q1_total_query = """
SELECT COUNT(*) AS total_q1_signups
FROM users
WHERE signup_date BETWEEN '2024-01-01' AND '2024-03-31';
"""

print(pd.read_sql(q1_total_query, engine).to_string(index=False))
```

## Q8 — The over-paying Premium/Family users

**Answer:** **7 users** are currently on Premium/Family plans while their watch history contains only Basic-tier shows. The current plan is taken from the most recent subscription active as of 30 June 2024. fileciteturn0file0L236-L249


```python
# Q8: The over-paying Premium/Family users

query = """
WITH current_subscriptions AS (
    SELECT
        s.*,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date DESC, subscription_id DESC
        ) AS rn
    FROM subscriptions s
    WHERE status = 'active'
      AND (end_date IS NULL OR end_date > '2024-06-30')
),
premium_family_users AS (
    SELECT user_id
    FROM current_subscriptions
    WHERE rn = 1
      AND plan IN ('Premium', 'Family')
)
SELECT COUNT(*) AS over_paying_users
FROM premium_family_users p
WHERE NOT EXISTS (
    SELECT 1
    FROM watch_sessions ws
    JOIN shows sh ON sh.show_id = ws.show_id
    WHERE ws.user_id = p.user_id
      AND sh.min_plan IN ('Premium', 'Family')
);
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q9 — Upgrade success cohort

**Answer:** **55 users** qualify. Their average time from signup to first Premium/Family upgrade is **64.96 days**. fileciteturn0file0L250-L262


```python
# Q9: Upgrade success cohort

query = """
WITH ordered_subscriptions AS (
    SELECT
        s.*,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY start_date, subscription_id
        ) AS rn
    FROM subscriptions s
),
first_subscription AS (
    SELECT user_id, plan AS first_plan
    FROM ordered_subscriptions
    WHERE rn = 1
),
first_upgrade AS (
    SELECT
        s.user_id,
        MIN(s.start_date) AS first_upgrade_date
    FROM subscriptions s
    JOIN first_subscription f ON f.user_id = s.user_id
    WHERE f.first_plan = 'Basic'
      AND s.plan IN ('Premium', 'Family')
    GROUP BY s.user_id
),
active_users AS (
    SELECT DISTINCT user_id
    FROM subscriptions
    WHERE status = 'active'
      AND (end_date IS NULL OR end_date > '2024-06-30')
)
SELECT
    COUNT(*) AS qualifying_users,
    ROUND(
        AVG(DATEDIFF(fu.first_upgrade_date, u.signup_date)),
        2
    ) AS avg_days_signup_to_first_upgrade
FROM first_upgrade fu
JOIN users u ON u.user_id = fu.user_id
JOIN active_users a ON a.user_id = fu.user_id
WHERE u.signup_date BETWEEN '2024-01-01' AND '2024-01-31';
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q10 — Cliffhanger comebacks

**Answer:** **4,345 cliffhanger comeback events** occurred. **S088 — Rayalaseema Raga** generated the most, with **64 events**. Each `(user, show, incomplete_date)` is counted once. fileciteturn0file0L263-L277


```python
# Q10: Cliffhanger comebacks

query = """
WITH comeback_events AS (
    SELECT DISTINCT
        first_session.user_id,
        first_session.show_id,
        first_session.session_date AS incomplete_date
    FROM watch_sessions first_session
    JOIN watch_sessions later_session
      ON later_session.user_id = first_session.user_id
     AND later_session.show_id = first_session.show_id
     AND later_session.session_date BETWEEN
         DATE_ADD(first_session.session_date, INTERVAL 1 DAY)
         AND DATE_ADD(first_session.session_date, INTERVAL 7 DAY)
    WHERE first_session.completed = 0
)
SELECT COUNT(*) AS total_comeback_events
FROM comeback_events;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

```python
# Q10: Show with the most cliffhanger comeback events.

top_show_query = """
WITH comeback_events AS (
    SELECT DISTINCT
        first_session.user_id,
        first_session.show_id,
        first_session.session_date AS incomplete_date
    FROM watch_sessions first_session
    JOIN watch_sessions later_session
      ON later_session.user_id = first_session.user_id
     AND later_session.show_id = first_session.show_id
     AND later_session.session_date BETWEEN
         DATE_ADD(first_session.session_date, INTERVAL 1 DAY)
         AND DATE_ADD(first_session.session_date, INTERVAL 7 DAY)
    WHERE first_session.completed = 0
)
SELECT
    ce.show_id,
    s.title,
    COUNT(*) AS comeback_events
FROM comeback_events ce
JOIN shows s ON s.show_id = ce.show_id
GROUP BY ce.show_id, s.title
ORDER BY comeback_events DESC, ce.show_id
LIMIT 1;
"""

print(pd.read_sql(top_show_query, engine).to_string(index=False))
```

## Q11 — Consecutive-week engagement

**Answer:** **1,675 users** have a streak of at least 4 consecutive ISO weeks. The longest streak is **26 weeks**; **U00213** is one user with that longest streak. The query uses the classic gaps-and-islands pattern described in the brief. fileciteturn0file0L282-L296


```python
# Q11: Consecutive-week engagement

query = """
WITH distinct_weeks AS (
    SELECT DISTINCT
        user_id,
        DATE_SUB(session_date, INTERVAL WEEKDAY(session_date) DAY) AS week_start
    FROM watch_sessions
    WHERE user_id IS NOT NULL
),
numbered_weeks AS (
    SELECT
        user_id,
        week_start,
        ROW_NUMBER() OVER (
            PARTITION BY user_id
            ORDER BY week_start
        ) AS week_number
    FROM distinct_weeks
),
streak_groups AS (
    SELECT
        user_id,
        DATE_SUB(week_start, INTERVAL week_number WEEK) AS streak_group
    FROM numbered_weeks
),
streak_lengths AS (
    SELECT
        user_id,
        streak_group,
        COUNT(*) AS streak_length
    FROM streak_groups
    GROUP BY user_id, streak_group
),
users_with_4_week_streak AS (
    SELECT DISTINCT user_id
    FROM streak_lengths
    WHERE streak_length >= 4
)
SELECT
    (SELECT COUNT(*) FROM users_with_4_week_streak)
        AS users_with_4_plus_week_streak,
    user_id AS example_longest_streak_user,
    streak_length AS longest_streak_weeks
FROM streak_lengths
ORDER BY streak_length DESC, user_id
LIMIT 1;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

## Q12 — Churn signal detection

**Answer:** **521 churn-signal users**. A user qualifies only when they had some May activity and their June watch minutes dropped by at least 50%. The aggregation is performed before filtering, following the CTE pattern recommended in the brief. fileciteturn0file0L297-L320


```python
# Q12: Churn signal detection

query = """
WITH monthly_watch AS (
    SELECT
        user_id,
        SUM(
            CASE
                WHEN session_date BETWEEN '2024-05-01' AND '2024-05-31'
                THEN watch_minutes ELSE 0
            END
        ) AS may_watch_minutes,
        SUM(
            CASE
                WHEN session_date BETWEEN '2024-06-01' AND '2024-06-30'
                THEN watch_minutes ELSE 0
            END
        ) AS june_watch_minutes
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date BETWEEN '2024-05-01' AND '2024-06-30'
    GROUP BY user_id
),
churn_signals AS (
    SELECT
        user_id,
        may_watch_minutes,
        june_watch_minutes,
        ROUND(
            100.0 * (may_watch_minutes - june_watch_minutes)
            / may_watch_minutes,
            2
        ) AS drop_percentage
    FROM monthly_watch
    WHERE may_watch_minutes > 0
      AND june_watch_minutes <= may_watch_minutes * 0.50
)
SELECT
    cs.user_id,
    u.name,
    cs.may_watch_minutes,
    cs.june_watch_minutes,
    cs.drop_percentage
FROM churn_signals cs
JOIN users u ON u.user_id = cs.user_id
ORDER BY cs.user_id;
"""

result = pd.read_sql(query, engine)
print(result.to_string(index=False))
```

```python
# Q12: Total number of churn-signal users.

churn_count_query = """
WITH monthly_watch AS (
    SELECT
        user_id,
        SUM(CASE WHEN session_date BETWEEN '2024-05-01' AND '2024-05-31'
                 THEN watch_minutes ELSE 0 END) AS may_watch_minutes,
        SUM(CASE WHEN session_date BETWEEN '2024-06-01' AND '2024-06-30'
                 THEN watch_minutes ELSE 0 END) AS june_watch_minutes
    FROM watch_sessions
    WHERE user_id IS NOT NULL
      AND session_date BETWEEN '2024-05-01' AND '2024-06-30'
    GROUP BY user_id
)
SELECT COUNT(*) AS total_churn_signal_users
FROM monthly_watch
WHERE may_watch_minutes > 0
  AND june_watch_minutes <= may_watch_minutes * 0.50;
"""

print(pd.read_sql(churn_count_query, engine).to_string(index=False))
```
